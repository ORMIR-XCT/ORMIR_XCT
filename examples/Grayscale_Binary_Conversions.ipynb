<a name="imports"></a>
## *Step 1: Imports:*

Import modules/packages and set the input image path. 

```python
import os
import sys 
from matplotlib import pyplot as plt
import numpy as np

import itk
import SimpleITK as sitk

import ITK_file_converter as ifc
from ormir_xct.util.sitk_itk import itk_sitk, sitk_itk
```

```python
# Set the main file that will be compared
gray_aim_read_path = "images/DYNACT2_214_TMC_MID.AIM"
gray_nifti_write_path = "images/test/gray_test.nii"

binary_nifti_read_path = "images/binary.nii"
binary_isq_write_path = "images/test/binary_test.isq"

# Pre-determined slice variable
slice_index = 10
```

<a name="grayscale"></a>
## *Grayscale AIM to NIFTI File Conversion*

Convert an AIM file to NIFTI file.

```python
# Convert the AIM file to NIFTI
ifc.file_read(gray_aim_read_path, gray_nifti_write_path)
```

```python
# Creating an ITK image from the AIM file
image = itk.imread(gray_aim_read_path)

# Create the array for the image
img_array = np.asarray(image)

# Slice the image
gray_aim_selected_slice = img_array[:, :, slice_index]
```

```python
# Creating an ITK image from the NIFTI file
image = itk.imread(gray_nifti_write_path)

# Create the array for the image
img_array = np.asarray(image)

# Slice the image
gray_nifti_selected_slice = img_array[:, :, slice_index]
```

```python
fig, axs = plt.subplots(1, 2, figsize=(15,15))

# AIM plot
im1 = axs[0].imshow(gray_aim_selected_slice, cmap='gray', )
axs[0].set_title(f"2D Slice # {slice_index} for AIM Image")
axs[0].set_xlabel("X-axis")
axs[0].set_ylabel("Y-axis")
fig.colorbar(im1, ax=axs[0], fraction=0.05).set_label("Pixel Intensity")

# NIFTI plot
im2 = axs[1].imshow(gray_nifti_selected_slice, cmap='gray')
axs[1].set_title(f"2D Slice # {slice_index} for NIFTI Image")
axs[1].set_xlabel("X-axis")
axs[1].set_ylabel("Y-axis")
fig.colorbar(im2, ax=axs[1], fraction=0.05).set_label("Pixel Intensity")

plt.tight_layout()
plt.show()
```

<a name="binary"></a>
## *Binary NIFTI to ISQ File Conversion*

Convet a NIFTI binary file to an ISQ binary file.

```python
# Convert the NIFTI file to ISQ
ifc.file_read(binary_nifti_read_path, binary_isq_write_path)
```

```python
# Creating an ITK image from the NIFTI file
image = itk.imread(binary_nifti_read_path)

# Create the array for the image
img_array = np.asarray(image)

# Slice the image
binary_nifti_selected_slice = img_array[:, :, slice_index]
```

```python
# Creating an ITK image from the ISQ file
image = itk.imread(binary_isq_write_path)

# Create the array for the image
img_array = np.asarray(image)

# Slice the image
binary_isq_selected_slice = img_array[:, :, slice_index]
```

```python
fig, axs = plt.subplots(1, 2, figsize=(15,15))

# ITK plot
im1 = axs[0].imshow(binary_nifti_selected_slice, cmap='gray', )
axs[0].set_title(f"2D Slice # {slice_index} for Binary NIFTI File")
axs[0].set_xlabel("X-axis")
axs[0].set_ylabel("Y-axis")
fig.colorbar(im1, ax=axs[0], fraction=0.05).set_label("Pixel Intensity")

# VTK plot
im2 = axs[1].imshow(binary_isq_selected_slice, cmap='gray')
axs[1].set_title(f"2D Slice # {slice_index} for Binary ISQ File")
axs[1].set_xlabel("X-axis")
axs[1].set_ylabel("Y-axis")
fig.colorbar(im2, ax=axs[1], fraction=0.05).set_label("Pixel Intensity")

plt.tight_layout()
plt.show()
```
